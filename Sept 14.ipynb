```python
num1 = float(input("Enter Number 1: "))
num2 = float(input("Enter Number 2: "))

print("1 is add, 2 is subtract, 3 is multiply and 4 is division.")

user_input = input("Input Operator Symbol: ")

if user_input == "+":
    print(num1+num2)
elif user_input == "-":
    print(num1-num2)
elif user_input == "*":
    print(num1*num2)
elif user_input == "/":
    try:
        print(num1/num2)
    except ZeroDivisionError:
        print("Division by Zero Error.")

else: print("Enter a valid Operator.") 
    
```
